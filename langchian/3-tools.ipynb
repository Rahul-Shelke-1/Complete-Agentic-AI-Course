## Tools

Models can request to call tools that perform tasks such as fetching data from a database, searching the web, or running code. Tools are pairing of:

1. A schema, including the name of the tool, description and/or argument definitions (often a JSON schema)
2. A function or coroutine to execute.

```python
import os
from dotenv import load_dotenv
from langchain.chat_models import init_chat_model

os.environ["GROQ_API_KEY"] = os.getenv("GROQ_API_KEY")
model = init_chat_model("groq:qwen/qwen3.6-27b")
respone = model.invoke("Why do parrots talk?")
respone
```
