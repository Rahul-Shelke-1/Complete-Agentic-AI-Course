## Streaming and Batch

### Streaming

Most models can stream their output content while it is being generated. By displaying output progresively. streaming significantly improves user experience, particularly for longer responses. Calling stream() returns an iterator that yields output chunks as they are processed. You can use loop to processes each chunk in real-time:

```python
model.invoke("Write me a 200 words paragraph on Artificial Intelligence")
```

```python
for chunk in model.stream("Write me a 200 words paragraph on Artificial Intelligence"):
    print(chunk.text, end="", flush=True)
```

### Batch

Batching a collection of independent requests to a model can significantly improve performance and reduce costs, as the processing can be done in parallel:

```python
responses = model.batch(
    [
        "Why do parrets have colorful feathers?",
        "How do airplanes fly?",
        "What is quantum computing?"
    ]
)

for response in responses:
    print(response) 
```

```python
responses = model.batch(
    [
        "Why do parrets have colorful feathers?",
        "How do airplanes fly?",
        "What is quantum computing?"
    ],
    config = {
        'max_currency': 5, # Limit to 5 parallel calles
    }
)

for response in responses:
    print(response) 
```
