## Messages

Messages are the fundamental unit of context for models in LangChain. They represent the input and the output of models, carrying both the content and metadata needed to represent the state of a conversation when intercating with an LLM. Messages are objects that contain:

- Role - Identifies the message type (e.g, system, user)
- Content - Represents the actual content of the message (like text, image, audio, documents etc)
- Metadata - Optional field such as response information, message IDs and token usage.

LangChain provides a standard message type that works across all model provides, ensuring the behaviour regardless od the model being called.

```python
import os
from dotenv import load_dotenv
from langchain.chat_models import init_chat_model
load_dotenv()

os.environ["GROQ_API_KEY"] = os.getenv("GROQ_API_KEY")
model = init_chat_model("groq:qwen/qwen3.6-27b")
```

```python
model.invoke("please tell me what is artifical intelligence ?")
```

## Message Prompts

Alternatively, you can pass the list messages to the model by providing a list of messages objects.

#### Message types:

- System message - Tell the model how to behave and provides the context for interactions
- Human message - Represents the user input and interactions with the model
- AI message - Responses generated by the model, including text content, tool calls, and metadata
- Tool message - Represents the outputs of tool calls

## System Message

A SystemMessage represent an initial set of instrcuctions that primes the model's behaviour. you can use a system message to set the tone, define the model's role, and establish guidelines for responses.

## Human Message

A HumanMessage represnts user input and interactions. They can contain text, images, audio, files and any other amount of multimodel content

## AI Message

An AIMessage represents the output of model invocation. They can include multimodel data, tool calls, and provider specific metadata that you can later access.

## Tool Message

For models that support tool calling, AI messages can contain tool calls. Tool messages are used to pass the results of a single tool execution back to the model.

```python
from langchain.messages import SystemMessage, HumanMessage, AIMessage

messages = [
    SystemMessage("You are a poerty expert"),
    HumanMessage("Write a poem on artificial intelligence"),
]

response = model.invoke(messages)
response.content
```

```python
for chunk in model.stream(messages, ):
    print(chunk.text, end="", flush=True)
```

```python
system_msg = SystemMessage("You are a helpful coding sssitent.")

messages = [
    system_msg,
    HumanMessage("How do I create REST API?")
]

response = model.invoke(messages)
print(response.content)
```

```python
## Detailed info to the LLM through system message
from langchain.messages import SystemMessage, HumanMessage

system_msg = SystemMessage("""
You are a senior python developer with experities in web frameworks.
Always provide code exmaples and explain your reasoning.
Be concise but thorough in your explaination.
""")

messages = [
    system_msg,
    HumanMessage("How do I create REST API?")
]

response = model.invoke(messages)
print(response.content)
```

```python
## Message Metadata

human_msg = HumanMessage(
    content="Hello!",
    name="alice", # optional : identify different users
    id="msg-123" # optional : unique identifier for tracing
)
```

```python
response = model.invoke([human_msg])

response
```

```python
from langchain.messages import HumanMessage, SystemMessage

# Create an AI message anually (e.g, for converstation history)
ai_msg = AIMessage("I'd be happy to help you with that question!")

# Add to converstation history
messages = [
    SystemMessage("Your a helpful assistent"),
    HumanMessage("Can you help me?"),
    ai_msg, # Insert as if it is from the model
    HumanMessage("Great! What's 2+2?")
]

response = model.invoke(messages)

print(response.content)
```

```python
response.usage_metadata
```

```python
from langchain.messages import AIMessage, ToolMessage

# After a model makes a tool call
# (Here, we demonstarte manually creating the messages for brevity)
ai_message = AIMessage(
    content = [],
    tool_calls = [{
        "name": "get_weather",
        "args": {"location": "San Fransisco"},
        "id": "call_123"
    }]
)

# Execute tool and create result message
weather_result = "Sunny, 72^oF"
tool_message = ToolMessage(
    content = weather_result,
    tool_call_id = "call_123" # Must match the call ID
)

# Continue conversation
messages = [
    HumanMessage("What's the weather in San Fransisco?"),
    ai_message, # Model's tool call
    tool_message, # Tool execution result
]

response = model.invoke(messages) # Model processes the model
```
